# Gradients

Heating and cooling **gradients** (rate of change per hour) of a temperature series, classified by direction and season.

All plots are **interactive** — hover to see exact values.

```python
import plotly.io as pio
pio.renderers.default = "notebook_connected"
```

## Load sample data

The bundled `flat_temp.csv` contains hourly indoor temperature readings.

```python
import pandas as pd
from importlib import resources

data_path = resources.files("pyedautils") / "data" / "flat_temp.csv"
temp = pd.read_csv(data_path, sep=";", index_col="timestamp", parse_dates=True)
temp.columns = ["temperature"]
temp.head()
```

## Compute gradients

`compute_gradients` resamples to hourly means, takes the step-to-step difference and keeps only steps above the magnitude `threshold`. The sign decides the direction (heating vs cooling).

```python
from pyedautils.data_prep.gradients import compute_gradients

grad = compute_gradients(temp["temperature"], threshold=0.3)
grad.head()
```

## Summary per direction

`summarize_gradients` gives mean / min / max and the count per direction.

```python
from pyedautils.data_prep.gradients import summarize_gradients

summarize_gradients(grad)
```

## Boxplots by season

`plot_gradient_boxplots` compares the heating and cooling spread per group (season, quarter, month, weekday or hour).

```python
from pyedautils.plots import plot_gradient_boxplots

fig = plot_gradient_boxplots(grad, groupby="season")
fig.show()
```
